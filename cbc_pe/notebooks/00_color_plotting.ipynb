## Palettes

```python
import matplotlib.pyplot as plt
import numpy as np

professional_palette = [
    "#1F77B4",  # azul
    "#FF7F0E",  # naranja
    "#2CA02C",  # verde
    "#D62728",  # rojo
    "#9467BD",  # morado
    "#8C564B",  # marrón
    "#E377C2",  # rosa
    "#7F7F7F",  # gris
    "#BCBD22",  # oliva
    "#17BECF",  # cian
]


executive_palette = [
    "#264653",
    "#2A9D8F",
    "#E9C46A",
    "#F4A261",
    "#E76F51",
    "#6D597A",
    "#355070",
    "#8AB17D",
]


science_palette = [
    "#0072B2",
    "#D55E00",
    "#009E73",
    "#CC79A7",
    "#F0E442",
    "#56B4E9",
    "#E69F00",
    "#000000",
]


hep_stack_palette = [
    "#4E79A7",  # azul
    "#F28E2B",  # naranja
    "#59A14F",  # verde
    "#E15759",  # rojo suave
    "#B07AA1",  # púrpura
    "#9C755F",  # marrón
    "#EDC948",  # amarillo
    "#76B7B2",  # turquesa
]


hep_line_palette = [
    "#0072B2",  # azul
    "#D55E00",  # vermellón
    "#009E73",  # verde
    "#CC79A7",  # púrpura
    "#000000",  # negro
    "#E69F00",  # naranja
    "#56B4E9",  # celeste
]
```

## Palette Enumeration

```python
from enum import Enum

class Palette(Enum):
    SCIENCE = science_palette
    EXECUTIVE = executive_palette
    HEP_STACK = hep_stack_palette
    HEP_LINE = hep_line_palette
    PROFESSIONAL = professional_palette


def set_palette(palette: Palette) -> None:
    plt.rcParams["axes.prop_cycle"] = plt.cycler(color=palette.value)
```

## Chosing colors

```python
set_palette(Palette.EXECUTIVE)
```

```python
# STACK

categorias = ["A", "B", "C", "D", "E", "F", "G", "H"]
valores = [12, 19, 7, 14, 22, 9, 16, 11]

for i in range(len(categorias)):
    plt.bar(categorias[i], valores[i])


plt.title("Barras usando executive_palette")
plt.xlabel("Categoría")
plt.ylabel("Valor")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
# LINES

x = np.linspace(0, 10, 200)

for i in range(8):
    y = np.sin(x + i * 0.4)
    plt.plot(x, y, linewidth=2.5, label=f"Serie {i + 1}")

plt.title("Visualización de executive_palette")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

## CMS/HEP Plotting Style

```python
plt.rcParams.update({
    "figure.figsize": (7, 5),
    "figure.dpi": 120,

    "axes.linewidth": 1.2,
    "axes.labelsize": 13,
    "axes.titlesize": 14,

    "xtick.labelsize": 11,
    "ytick.labelsize": 11,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.top": True,
    "ytick.right": True,

    "legend.frameon": False,
    "legend.fontsize": 10,

    "grid.alpha": 0.25,
    "lines.linewidth": 2,
})
```

```python
x = np.linspace(0, 10, 500)

for i in range(6):
    y = np.exp(-0.15 * x) * np.sin(x + i * 0.6)
    plt.plot(x, y, linewidth=2, label=f"Modelo {i+1}")

plt.xlabel(r"$p_T$ [GeV]")
plt.ylabel(r"Eventos normalizados")
plt.title("Ejemplo de paleta científica")
plt.grid(alpha=0.3)
plt.legend(frameon=False)
plt.tight_layout()
plt.show()
```
